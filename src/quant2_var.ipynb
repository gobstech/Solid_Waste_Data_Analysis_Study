Este notebook foi criado para importar a tabela de Resíduos Sólidos em Excel para o ambiente do Google Colab. Aqui mostraremos a importação da tabela, assim como a criação das tabelas.

```python
# Importação das bibliotecas que utilizaremos
import pandas as pd
import math
```

```python
# Importação do arquivo de tabela a ser utilizado, através do path na seção "Files" do Colab, contendo as informações de resíduos sólidos
file='./Solid_waste_pt.xlsx'
```

```python
# Visão geral da tabela
df=pd.read_excel(file)
df
```

```python
# Tabela que mostra a quantidade de coletas realizadas em cada regiao
df_Poder_Calorifico_MJkg=df.Poder_Calorifico_MJkg.value_counts()
df_Poder_Calorifico_MJkg
```

```python
# Ordenação da tabela em ordem alfabética
df_Poder_Calorifico_MJkg=df.Poder_Calorifico_MJkg.value_counts().sort_index()
df_Poder_Calorifico_MJkg
```

```python
# Impressão do total das coletas realizadas
total_df_Poder_Calorifico_MJkg=df_Poder_Calorifico_MJkg.sum()
total_df_Poder_Calorifico_MJkg
```

```python
# Impressão da tabela com as Regiões de Coleta ordenadas de forma alfabética
dados_v_Poder_Calorifico_MJkg=pd.DataFrame(df_Poder_Calorifico_MJkg)
dados_v_Poder_Calorifico_MJkg
```

```python
# @title
# Reinicia os índices da região de coleta
dados_v_Poder_Calorifico_MJkg=pd.DataFrame(df_Poder_Calorifico_MJkg).reset_index()
# Renomeia as colunas na tabela com o total de resíduos coletados em cada região da cidade
dados_v_Poder_Calorifico_MJkg.columns=['Poder Calorífico','Total']
dados_v_Poder_Calorifico_MJkg

print(dados_v_Poder_Calorifico_MJkg)
```

```python
# Cálculo da frequência relativa
dados_v_Poder_Calorifico_MJkg['Frequência relativa(%)']=round((dados_v_Poder_Calorifico_MJkg['Total']/total_df_Poder_Calorifico_MJkg)*100,2)
print(dados_v_Poder_Calorifico_MJkg)
```

```python
tabela_Poder_Calorifico_MJkg=pd.DataFrame(dados_v_Poder_Calorifico_MJkg)
tabela_Poder_Calorifico_MJkg
```
