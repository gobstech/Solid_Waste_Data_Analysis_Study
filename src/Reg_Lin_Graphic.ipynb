```python
#Importando as bibliotecas

import pandas as pd
import math
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy import stats
from scipy.stats import linregress
```

```python
#Definindo a variável "arquivo" pelo nome da planilha.

arquivo='Solid_waste_pt.xlsx'
```

```python
#lendo a planilha no Python. Escolhemos o nome df para armazenar.

df=pd.read_excel(arquivo)
```

```python
df=df.dropna()
```

```python
df
```

```python
x=df.Umidade_Residuo_pct.values
y=df.Poder_Calorifico_MJkg.values
```

```python
#Plotando o diagrama de dispersão de x por y.
plt.plot(x, y, 'o',label="dados")
#note que o nome do Diagrama é sempre y por x
plt.title('Diagrama de Dispersão: Poder Calorifíco - MJ/Kg x Umidade Resíduo - %')
plt.xlabel("Umidade Resíduo - %")
plt.ylabel("Poder Calorifíco - MJ/Kg")
plt.legend()
plt.show()
```

```python
res = stats.linregress(x, y)
```

```python
res
```

```python
res.slope
```

```python
plt.plot(x, y, 'o', label='dados')
plt.plot(x, res.slope*x + res.intercept, 'r', label='reta de regressão')
plt.title('Regressão Linear Simples: Poder Calorifíco - MJ/Kg x Umidade Resíduo - %')
plt.xlabel("Umidade Resíduo - %")
plt.ylabel('Poder Calorifíco - MJ/Kg')
#cada grupo ajusta esses limites de acordo com seus dados. Use números inteiros
minimox=df.Umidade_Residuo_pct.min()
maximox=df.Umidade_Residuo_pct.max()
plt.xlim(minimox,maximox)
plt.legend()
plt.show()
```

```python
print(f"Coeficiente de correlação r = {res.rvalue:.4f}")
print( f"Coeficiente linear =  {res.intercept:.4f}")
print( f"Coeficiente angular =  {res.slope:.4f}")
print(f"Reta de regressão linear y = {res.slope:.4f}",f"*x + {res.intercept:.4f}" )
print(f"Coeficiente de determinação r^2 = {res.rvalue**2:.4f}")
```
